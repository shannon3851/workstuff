### **UPDATED - res2 analysis**

```python
# RESIDENTIAL 2 dwelling unit density
# Read-in data
import pandas as pd

file_path = '/content/RES2_version3_datatable.xls'
res2 = pd.read_excel(file_path)
print(res2.head())
```

```python
# check total rows
len(res2)
```

```python
# Number of parcels below 6du/acre
res2_count = (res2['Density']<6).sum()
print(res2_count)
```

```python
# % of parcels below 6du/acre
res2percentage = (res2_count / 4100)*100
print(res2percentage)
```

```python
# Number of parcels above 6du/acre
res2_count2 = (res2['Density']>6).sum()
print(res2_count2)
```

```python
#% of parcels above 6du/acre
res2percentage2 = (res2_count2 / 4100)*100
print(res2percentage2)
```

```python
# equal to 6
res2_count3 = (res2['Density']==6).sum()
print(res2_count3)
```
