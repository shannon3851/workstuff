### **RES2 - POTENTIAL DENSITY 8DU/ACRE**

```python
import pandas as pd

file_path = '/content/RES2_version3_datatable.xls'
res2 = pd.read_excel(file_path)
print(res2.head(2))
print(len(res2))
```

```python
print(res2['Density'].describe())
```

```python
# number parcels below 8du/acre
below = (res2['Density']<=8).sum()
print('below', below)

# % parcels below 8du/acre
below_percent = ((below/4100)*100)
print('percent below:', below_percent)
```

```python
# parcels above 8du/acre
above = (res2['Density']>=8).sum()
print('above', above)

#percent above 8du/acre
above_percent = ((above/4100)*100)
print('percent above:', above_percent)
```

```python
# acres below 8du/acre
acres_below = res2.loc[res2['Density']<=8, 'Acreage'].sum()
print('acres below:', acres_below)

#acres above 8du/acre
acres_above = res2.loc[res2['Density']>=8, 'Acreage'].sum()
print('acres above:', acres_above)
```

```python
# units below 8du/acre
units_below = res2.loc[res2['Density']<=8, 'Join_Count'].sum()
print('units below:', units_below)


# units above 8du/acre
units_above = res2.loc[res2['Density']>=8, 'Join_Count'].sum()
print('units above:', units_above)
```

```python
print(res2['Join_Count'].sum())
```

```python
# export parcels below 8du/acre
acres_below_export = res2.loc[res2['Density']<=8]

print(acres_below_export)

acres_below_export.to_excel('R2_below8du.xlsx', index=False)


```

```python
# export parcels above 8du/acre
acres_above_export = res2.loc[res2['Density']>=8]
print(acres_below_export)

acres_above_export.to_excel('R2_above8du.xlsx', index=False)
```
