###**UNIT DENSITY PER PARISH**

```python
# UNIT DENSITY PER PARISH
```

```python
import pandas as pd

file_path = '/content/V2_ResidentialUnit_Joins_TableToExcel.xlsx'
df = pd.read_excel(file_path)
print(df.head())
```

```python
# group by parish
result = (
    df.groupby('Parish_Name')['Units']
    .sum()
    .reset_index()
    .sort_values('Parish_Name')

)
print(result)
```

```python
# export to excel
result.to_excel('units_by_parish.xlsx', index=False)
```

```python
# group by residential 1&2
result = (
    df.groupby('Predomin_1')['Units']
    .sum()
    .reset_index()
    .sort_values('Predomin_1')
)
print(result)
```

```python
# group by parish, unit, res
result = (
    df[df['Predomin_1'].isin(['Residential 1', 'Residential 2'])]
       .pivot_table(
           index='Parish_Name',
           columns='Predomin_1',
           values='Units',
           aggfunc='sum',
           fill_value=0
       )
       .reset_index()
       .sort_values('Parish_Name')
       )
print(result)

result.to_excel('residential_units_by_parish.xlsx', index=False)
```
