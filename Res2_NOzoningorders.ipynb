```python
# lots below & above 6du/acre
```

```python
below6du = (res2['Density']<=6).sum()
print('lots below 6du:', below6du)

percentbelow6du  = ((below6du/len(res2))*100)
print('percent below 6du:', percentbelow6du)
```

```python
above6du = (res2['Density']>=6).sum()
print('lots above 6du:', above6du)

percentover6du  = ((above6du/len(res2))*100)
print('percent above 6du:', percentover6du)
```

```python
# distribution
print(res2['Density'].describe())

print('median:', res2['Density'].median())
```

```python
# acres below & above 6du/acre
```

```python
acresbelow6du = res2.loc[res2['Density']<=6, 'Acreage'].sum()
print('acres below 6du:', acresbelow6du)

percentacresbelow = ((acresbelow6du/res2['Acreage'].sum())*100)
print('percent acres below 6du:', percentacresbelow)
```

```python
acresabove6du = res2.loc[res2['Density']>=6, 'Acreage'].sum()
print('acres above 6du:', acresabove6du)

percentacresabove = ((acresabove6du/res2['Acreage'].sum())*100)
print('percent acres above 6du:', percentacresabove)
```

```python
# units below and above 6du/acre
```

```python
unitsbelow6du = res2.loc[res2['Density']<=6, 'Join_Count'].sum()
print('units below 6du:', unitsbelow6du)

percentunitsbelow = ((unitsbelow6du/res2['Join_Count'].sum())*100)
print('percent units below 6du:', percentunitsbelow)
```

```python
unitsabove6du = res2.loc[res2['Density']>=6, 'Join_Count'].sum()
print('units above 6du:', unitsabove6du)

percentunitsabove = ((unitsabove6du/res2['Join_Count'].sum())*100)
print('percent units above 6du:', percentunitsabove)
```
